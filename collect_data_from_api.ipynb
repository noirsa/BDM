[Weather Image Recognition ](https://www.kaggle.com/datasets/jehanbhathena/weather-dataset) is a dataset of images from kaggle. We apply this dataset because it covers rare or specific weather conditions.

```
import kagglehub
import torch
from torchvision import datasets, transforms
from torch.utils.data import DataLoader
# Download latest version
path = kagglehub.dataset_download("jehanbhathena/weather-dataset")
import os
from PIL import Image

def check_landing_zone(root_path):
    valid_images = 0
    corrupted_files = []

    categories = [d for d in os.listdir(root_path) if os.path.isdir(os.path.join(root_path, d))]

    print(f"Checking {len(categories)} weather categories...")

    for category in categories:
        category_path = os.path.join(root_path, category)
        files = os.listdir(category_path)

        for file in files:
            file_path = os.path.join(category_path, file)
            try:
                # We only open the image header, not the whole pixel data (saves RAM)
                with Image.open(file_path) as img:
                    img.verify()
                valid_images += 1
            except (IOError, SyntaxError) as e:
                print(f"CORRUPT: {file_path}")
                corrupted_files.append(file_path)

    print("--- Scan Complete ---")
    print(f"Total Valid Images: {valid_images}")
    print(f"Total Corrupted: {len(corrupted_files)}")
    return corrupted_files
check_landing_zone(path+"/dataset")
```

```
import urllib.request
import xml.etree.ElementTree as ET
import io
from pypdf import PdfReader

# 1. Search for climate documents
url = 'http://export.arxiv.org/api/query?search_query=all:climate+change&max_results=1'

# 2. Get the Metadata (XML)
with urllib.request.urlopen(url) as response:
    xml_data = response.read().decode('utf-8')

# 3. Parse XML to find the PDF link
root = ET.fromstring(xml_data)
ns = {'atom': 'http://www.w3.org/2005/Atom'}
entry = root.find('atom:entry', ns)

if entry:
    title = entry.find('atom:title', ns).text.strip()
    pdf_url = entry.find("atom:link[@title='pdf']", ns).attrib['href']

    print(f"--- PAPER TITLE: {title} ---")
    print(f"--- FETCHING CONTENT FROM: {pdf_url} ---\n")

    # 4. Fetch the PDF directly into memory (No file saved)
    with urllib.request.urlopen(pdf_url) as pdf_response:
        pdf_file = io.BytesIO(pdf_response.read())
        reader = PdfReader(pdf_file)

        # 5. Print text from the first page as a check
        for text in reader.pages:
          print(text.extract_text()) # Print first 1000 characters
        print("\n--- END OF PREVIEW ---")
else:
    print("No documents found.")
```

```
import asyncio
from playwright.async_api import async_playwright

async def scrape_preventionweb():
    async with async_playwright() as p:
        # Launch browser
        browser = await p.chromium.launch(headless=True)
        page = await browser.new_page()

        # Target URL
        url = "https://www.preventionweb.net/search#query=Global%20Natural%20Disaster%20Assessment%20report"

        print(f"Opening {url}...")

        try:
            # Use domcontentloaded which is faster and less prone to network timeouts
            await page.goto(url, wait_until="domcontentloaded", timeout=60000)

            # Manually wait for the specific results container with a generous timeout
            print("Waiting for search results to load...")
            await page.wait_for_selector('.search-result', timeout=30000)

            # Extract titles and links
            results = await page.query_selector_all('.search-result')

            print(f"\nFound {len(results)} reports:\n")
            for result in results:
                title_element = await result.query_selector('h3 a')
                if title_element:
                    title = await title_element.inner_text()
                    link = await title_element.get_attribute('href')
                    full_link = f"https://www.preventionweb.net{link}" if link.startswith('/') else link
                    print(f"Title: {title}")
                    print(f"URL: {full_link}\n")

        except Exception as e:
            print(f"An error occurred: {e}")
            # Fallback debug check
            content = await page.content()
            if "search-result" not in content:
                print("Could not find search results in the current page state. The site may have changed its layout or blocked the request.")

        await browser.close()

await scrape_preventionweb()
```
